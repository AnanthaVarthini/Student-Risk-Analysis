Defining Risk criteria based on student grades.

```python
data['Academic_Performance_Index'] = (
    0.30 *data['Attendance (%)'] +
    0.30 * data['Midterm_Score'] +
    0.25 * data['Final_Score'] +
    0.15 * data['Assignments_Avg']
)
```

```python
def assign_risk_class(score):
    if score >= 75.0:
        return 'LOW'
    elif score >= 60.0:
        return 'MEDIUM'
    else:
        return 'HIGH'

data['Risk'] = data['Academic_Performance_Index'].apply(assign_risk_class)
```

```python
data = data.dropna(
    subset=["Risk"]
).reset_index(drop=True)
```

Training the model using the given dataset.

 Selected relevant features for predicting academic risk.

```python
selected_features = [
    "Gender",
    "Age",
    "Department",
    "test_preparation_course",
    "Attendance (%)",
    "Midterm_Score",
    "Final_Score",
    "Assignments_Avg",
    "Quizzes_Avg",
    "Participation_Score",
    "Projects_Score",
    "math_score",
    "reading_score",
    "writing_score",
    "science_score"
]


selected_features = [
    column
    for column in selected_features
    if column in data.columns
]
```

```python
X = data[selected_features].copy()
y = data["Risk"]

print("Number of features:", len(selected_features))
print("Features:")
print(selected_features)
```

```python
print("X:", X.shape)
print("y:", y.shape)
```

```python
numerical_features = (
    X.select_dtypes(
        include=["int64", "float64"]
    ).columns.tolist()
)

categorical_features = (
    X.select_dtypes(
        include=["object", "category"]
    ).columns.tolist()
)

print("\nNumerical Features:")
print(numerical_features)

print("\nCategorical Features:")
print(categorical_features)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)

print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import OneHotEncoder
```

This handles the missing values after the split, using information from the training data only.

```python
numerical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
```

test_preparation_course is categorical, so we convert its categories into numerical values using One-Hot Encoding.

```python
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])
```

Combining both,We get.

```python
preprocessor = ColumnTransformer([
    (
        "numerical",
        numerical_pipeline,
        numerical_features
    ),
    (
        "categorical",
        categorical_pipeline,
        categorical_features
    )
])
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

import warnings
warnings.filterwarnings("ignore")
```

Now ML part , we are using three different models.

```python
logistic_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(
        max_iter=2000,
        class_weight="balanced"
    ))
])
```

```python
logistic_model.fit(X_train,y_train)
```

```python
logistic_predictions = logistic_model.predict(X_test)
```

```python
print(classification_report(y_test, logistic_predictions))
```

```python
logistic_accuracy = accuracy_score(
    y_test,
    logistic_predictions
)

logistic_precision = precision_score(
    y_test,
    logistic_predictions,
    average="weighted"
)

logistic_recall = recall_score(
    y_test,
    logistic_predictions,
    average="weighted"
)

logistic_f1 = f1_score(
    y_test,
    logistic_predictions,
    average="weighted"
)

print("Logistic Regression")
print("Accuracy :", logistic_accuracy)
print("Precision:", logistic_precision)
print("Recall   :", logistic_recall)
print("F1 Score :", logistic_f1)
```

```python
decision_tree_model = Pipeline(
    [
    ("preprocessor", preprocessor),
    (
        "model",
        DecisionTreeClassifier
         (
            max_depth=6,
            class_weight="balanced",
            random_state=42
        )
    )
    ]
    )
```

```python
decision_tree_model.fit(
    X_train,
    y_train
)
```

```python
decision_tree_predictions = decision_tree_model.predict(
    X_test
)
```

```python
dt_accuracy = accuracy_score(
    y_test,
    decision_tree_predictions
)

dt_precision = precision_score(
    y_test,
    decision_tree_predictions,
    average="weighted"
)

dt_recall = recall_score(
    y_test,
    decision_tree_predictions,
    average="weighted"
)

dt_f1 = f1_score(
    y_test,
    decision_tree_predictions,
    average="weighted"
)

print("Decision Tree")
print("Accuracy :", dt_accuracy)
print("Precision:", dt_precision)
print("Recall   :", dt_recall)
print("F1 Score :", dt_f1)
```

```python
random_forest_model = Pipeline([
    ("preprocessor", preprocessor),
    (
        "model",
        RandomForestClassifier(
            n_estimators=300,
            max_depth=10,
            class_weight="balanced",
            random_state=42,
            n_jobs=-1
        )
    )
])
```

```python
random_forest_model.fit(
    X_train,
    y_train
)
```

```python
random_forest_predictions = random_forest_model.predict(
    X_test
)
```

```python
rf_accuracy = accuracy_score(
    y_test,
    random_forest_predictions
)

rf_precision = precision_score(
    y_test,
    random_forest_predictions,
    average="weighted"
)

rf_recall = recall_score(
    y_test,
    random_forest_predictions,
    average="weighted"
)

rf_f1 = f1_score(
    y_test,
    random_forest_predictions,
    average="weighted"
)

print("Random Forest")
print("Accuracy :", rf_accuracy)
print("Precision:", rf_precision)
print("Recall   :", rf_recall)
print("F1 Score :", rf_f1)
```

```python
results = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Decision Tree",
        "Random Forest"
    ],

    "Accuracy": [
        logistic_accuracy,
        dt_accuracy,
        rf_accuracy
    ],

    "Precision": [
        logistic_precision,
        dt_precision,
        rf_precision
    ],

    "Recall": [
        logistic_recall,
        dt_recall,
        rf_recall
    ],

    "F1 Score": [
        logistic_f1,
        dt_f1,
        rf_f1
    ]
})

results
```

```python
best_model_name = results.loc[
    results["F1 Score"].idxmax(),
    "Model"
]

print("Best Model:", best_model_name)
```

```python
def generate_recommendation(attendance, risk):

    if risk == "HIGH":

        if pd.notna(attendance) and attendance < 60:
            return (
                "Improve attendance immediately and "
                "focus on upcoming assessments."
            )

        else:
            return (
                "Focus strongly on upcoming assessments "
                "and seek academic support."
            )

    elif risk == "MEDIUM":

        if pd.notna(attendance) and attendance < 75:
            return (
                "Improve attendance consistency and "
                "focus on upcoming assessments."
            )

        else:
            return (
                "Maintain attendance and improve weaker "
                "academic areas."
            )

    else:

        return (
            "Maintain current attendance and "
            "academic performance."
        )
```
